```python
import pandas as pd
import numpy as np

fixed_entries = [
    {"question": "what is the annual fee",
     "answer": "The annual fee is Rs 500.",
     "keywords": "fee cost price charge",
     "category": "billing"},
    {"question": "how to reset password",
     "answer": "Go to Settings > Reset Password.",
     "keywords": "password reset login",
     "category": "account"},
    {"question": "what are your working hours",
     "answer": "We are open 9 AM to 5 PM.",
     "keywords": "hours timing open time",
     "category": "general"},
    {"question": "how can i pay the fee",
     "answer": "You can pay via UPI, card, or net banking.",
     "keywords": "pay payment upi fee",
     "category": "billing"},
]

roll = 1024170305
d = [int(digit) for digit in str(roll)[-2:]]
cat = ["billing", "account", "general"]
cat_main = []
for i in d:
    cat_main.append(cat[i % 3])
questions = [
    "how do i check my payment status",
    "how can i contact support"
]
answers = [
    "You can check your payment status in the Payments section of your account.",
    "You can contact support through the Help Center or email us."
]
keywords = [
    "payment status billing transaction",
    "support help contact"
]
newFqa = []
for i in range(2):
    newFqa.append({
        "question": questions[i],
        "answer": answers[i],
        "keywords": keywords[i],
        "category": cat_main[i]
    })
newFqa.extend(fixed_entries)


df = pd.DataFrame(newFqa)


print(df)
```

Q2

```python
def scoring(query,df):
  questions = [sent.lower().split() for sent in df['question']]
  actual = query.lower().split()
  order = []
  for i in range(len(questions)):
    cnt = 0
    for w in actual:
      if(w in questions[i]):
        cnt+=1
    order.append(cnt)
  sorted_order = np.argsort(order)
  ans =[]
  for i in range(len(order)):
    ans.append(df['question'][sorted_order[i]])
  return ans
```

```python
print("Please enter a query")
query = input()
ranked_FAQ = scoring(query,df)
print("Here are FAQ's in sorted order -> (increasing order of matching)")
print(ranked_FAQ)
```
