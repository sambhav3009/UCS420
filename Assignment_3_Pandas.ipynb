## Q6. Create `employees.csv`

The following data is exactly the sample dataset shown in the assignment.

```python
employee_data = {"Employee_ID": [101,102,103,104,105],
    "Name": ["Alice","Bob","Charlie","Diana","Edward"],
    "Department": ["HR","IT","IT","Marketing","Sales"],
    "Age": [29,34,41,28,38],
    "Salary": [50000,70000,65000,55000,60000],
    "Years_of_Experience": [4,8,10,3,12],
    "Joining_Date": ["2020-03-15","2017-07-19","2013-06-01","2021-02-10","2010-11-25"],
    "Gender": ["Female","Male","Male","Female","Male"],
    "Bonus": [5000,7000,6000,4500,5000],
    "Rating": [4.5,4.0,3.8,4.7,3.5]
}
employees_df = pd.DataFrame(employee_data)
employees_df["Joining_Date"]=pd.to_datetime(employees_df["Joining_Date"])
employees_df.to_csv("employees.csv", index=False)
employees_df
```

```python
print("First 5 rows:")
display(employees_df.head(5))

print("Last 3 rows:")
display(employees_df.tail(3))
```

```python
print("Average salary:", employees_df["Salary"].mean())
print("Total bonus:", employees_df["Bonus"].sum())
print("Youngest employee's age:", employees_df["Age"].min())
print("Highest performance rating:", employees_df["Rating"].max())
```

```python
employees_df.sort_values(by="Salary", ascending=False)
```

```python
def performance_category(rating):
    if rating >= 4.5:
        return "Excellent"
    elif rating >= 4.0:
        return "Good"
    return "Average"

employees_df["Performance_Category"] = employees_df["Rating"].apply(performance_category)
employees_df
```

```python
employees_df.isnull().sum()
```

```python
employees_df.rename(columns={"Employee_ID":"ID"}, inplace=True)
employees_df
```

```python
print("More than 5 years of experience:")
display(employees_df[employees_df["Years_of_Experience"] > 5])

print("IT department:")
display(employees_df[employees_df["Department"] == "IT"])
```

```python
employees_df["Tax"] = employees_df["Salary"] * 0.10
employees_df
```

```python
employees_df.to_csv("employees_modified.csv", index=False)
print("Saved as employees_modified.csv")
```
