```python
Sambhav= np.linspace(10, 100, 25)
print(Sambhav)
print("Dimensions:", Sambhav.ndim)
print("Shape:", Sambhav.shape)
print("Total elements:", Sambhav.size)
print("Data type:", Sambhav.dtype)
print("Total bytes:", Sambhav.nbytes)
trans=Sambhav.reshape(25,1)
print(trans)
print(Sambhav.T)
```

```python
ucs420_Sambhav = np.array([
    [10, 20, 30, 40],
    [50, 60, 70, 80],
    [90, 15, 20, 35]
])

print("Original Array:")
print(ucs420_Sambhav)

print("Mean:", np.mean(ucs420_Sambhav))
print("Median:", np.median(ucs420_Sambhav))
print("Maximum:", np.max(ucs420_Sambhav))
print("Minimum:", np.min(ucs420_Sambhav))
print("Unique elements:", np.unique(ucs420_Sambhav))

reshaped_ucs420_Sambhav = ucs420_Sambhav.reshape(4, 3)
print("Reshaped Array (4 x 3):")
print(reshaped_ucs420_Sambhav)

resized_ucs420_Sambhav = np.resize(ucs420_Sambhav, (2, 3))
print("Resized Array (2 x 3):")
print(resized_ucs420_Sambhav)
```
